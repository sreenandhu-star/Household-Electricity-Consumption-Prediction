```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# Load dataset
file_path = "/content/household_power_consumption.txt"

df = pd.read_csv(
    file_path,
    sep=";",
    na_values=["?"],
    low_memory=False
)

print("Dataset Loaded Successfully!")
print("\nFirst 5 Records:")
print(df.head())

print("\nDataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
# Convert electricity consumption to numeric
df["Global_active_power"] = pd.to_numeric(
    df["Global_active_power"],
    errors="coerce"
)

# Check missing values
print("Missing Values Before Processing:")
print(df["Global_active_power"].isnull().sum())

# Fill missing values using interpolation
df["Global_active_power"] = df["Global_active_power"].interpolate(
    method="linear"
)

# Remove any remaining missing values
df = df.dropna(subset=["Global_active_power"])

# Select first 10,000 consecutive records
df = df.iloc[:10000].copy()

print("\nMissing Values After Processing:")
print(df["Global_active_power"].isnull().sum())

print("\nSelected Dataset Shape:", df.shape)
print(df.head())
```

```python
# Prepare electricity consumption values
data = df[["Global_active_power"]].values

# Split chronologically: first 80% for training
split_index = int(len(data) * 0.8)

train_data = data[:split_index]
test_data = data[split_index:]

# Fit scaler only on training data
scaler = MinMaxScaler()

train_scaled = scaler.fit_transform(train_data)

test_scaled = scaler.transform(test_data)

# Combine scaled data for sequence creation
scaled_data = np.vstack((train_scaled, test_scaled))

print("Training Records:", len(train_data))
print("Testing Records:", len(test_data))
print("Normalization Completed!")
```

```python
sequence_length = 60

X_train = []
y_train = []

X_test = []
y_test = []

# Create training sequences
for i in range(sequence_length, len(train_scaled)):
    X_train.append(train_scaled[i-sequence_length:i])
    y_train.append(train_scaled[i, 0])

# Create testing sequences
# Use the previous 60 records, including the training boundary
for i in range(len(train_scaled), len(scaled_data)):
    X_test.append(scaled_data[i-sequence_length:i])
    y_test.append(scaled_data[i, 0])

X_train = np.array(X_train)
y_train = np.array(y_train)

X_test = np.array(X_test)
y_test = np.array(y_test)

print("X_train Shape:", X_train.shape)
print("y_train Shape:", y_train.shape)

print("X_test Shape:", X_test.shape)
print("y_test Shape:", y_test.shape)
```

```python
model = Sequential([
    LSTM(
        50,
        return_sequences=True,
        input_shape=(60, 1)
    ),

    Dropout(0.2),

    LSTM(50),

    Dropout(0.2),

    Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)

model.summary()
```

```python
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    y_train,
    epochs=20,
    batch_size=32,
    validation_split=0.1,
    shuffle=False,
    callbacks=[early_stop],
    verbose=1
)
```

```python
# Predict test data
predictions = model.predict(X_test)

# Convert values back to original scale
y_test_actual = scaler.inverse_transform(
    y_test.reshape(-1, 1)
)

predicted_actual = scaler.inverse_transform(
    predictions
)

# Calculate evaluation metrics
mae = mean_absolute_error(
    y_test_actual,
    predicted_actual
)

rmse = np.sqrt(
    mean_squared_error(
        y_test_actual,
        predicted_actual
    )
)

print("Model Evaluation Results")
print("------------------------")
print("MAE:", round(mae, 4), "kW")
print("RMSE:", round(rmse, 4), "kW")
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    y_test_actual[:300],
    label="Actual Consumption"
)

plt.plot(
    predicted_actual[:300],
    label="Predicted Consumption"
)

plt.title("Household Electricity Consumption Prediction")
plt.xlabel("Test Record")
plt.ylabel("Global Active Power (kW)")

plt.legend()
plt.grid(True)
plt.show()
```

```python
# Get the latest 60 measurements
last_60 = scaled_data[-60:]

# Reshape for LSTM input
last_60 = last_60.reshape(1, 60, 1)

# Predict next minute
next_prediction = model.predict(last_60, verbose=0)

# Convert to original scale
next_prediction = scaler.inverse_transform(
    next_prediction
)

print(
    "Predicted Next-Minute Electricity Consumption:",
    round(float(next_prediction[0, 0]), 4),
    "kW"
)
```
